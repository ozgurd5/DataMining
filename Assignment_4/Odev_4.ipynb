```python
#16. Python’da “seaborn” kütüphanesinden “mpg.csv” veri setini indiriniz, veri seti bilgilerini yazdırınız (info), veri setinin ilk 10 satrını ekrana yazdırınız

data = sns.load_dataset('mpg')

print(data.info())
print(data.head(10))
```

```python
#17. veri setinin uygun atributlraına göre ortalama, varyans, min, max, vs. değerleri yazdırınız,

data.describe()
```

```python
#18. “horsepower” ve “weight” atributları için, “origin” kırılımına (hue) göre “scatter plot” grafiğini oluşturunuz

plt.figure(figsize=(10,6))
sns.scatterplot(x=data['horsepower'], y=data['weight'], hue=data['origin'])
```

```python
#19. “mpg” ve “acceleration” değişkenleri arasında Kendall korelasyon katsayısını hesaplamalı ve yazdırmalı.

kendall = data[['mpg', 'acceleration']].corr(method='kendall')
print(f'Kendall Korelasyonu:\n{kendall}')
```

```python
#20. “horsepower”, “acceleration” ve “weight” değişkenlerine göre “mpg” değerinin tahmini için LinearRegression() modeli oluşturunuz. Modelin intercept ve katsayılarını yazdırınız
#NOT: kayıp veriler olabilir. Gerekirse kayıp verileri siliniz.

print("\nEksik değerlerin sayısı:\n", data.isnull().sum())
data = data.dropna(subset=["horsepower"]) # 6 adet eksik horsepower değeri mevcut, kaldırılıyor.

X_train, X_test, y_train, y_test = train_test_split(data.drop(columns=['mpg']), data['mpg'], test_size=0.1, random_state=300)

X_train = X_train[['horsepower', 'acceleration', 'weight']]
X_test = X_test[['horsepower', 'acceleration', 'weight']]

linearRegresyon = LinearRegression()
linearRegresyon.fit(X_train, y_train)

print(f'\nIntercept: {linearRegresyon.intercept_}\n' + f'Katsayılar: {linearRegresyon.coef_}')
```

```python
#21. Oluşturulan modelin R-squared değerini yazdırınız

print(f'R-squared: {linearRegresyon.score(X_test, y_test)}')
```

```python
#22. horsepower = 130, acceleration=13, weight=3500 olan bir otomobilin “mpg” değerini tahmin ediniz.

tahmin = linearRegresyon.predict([[130, 13, 3500]])
print(f'Tahmin: {tahmin}')
```

```python
#23. “mpg.csv” veri setini, “mpg”, “displacement”,”horsepower”,”weight”  ve “acceleration” değişkenlerine göre “origin” değerinin tahmini için %20 test,%80 eğitim oranında ayrıştırınız. Eğitim setini kullanarak RandomForest sınıflandırma modeli oluşturunuz

X = data[['mpg', 'displacement', 'horsepower', 'weight', 'acceleration']]
y = data['origin']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=300)

randomForest = RandomForestClassifier()
randomForest.fit(X_train, y_train)
```

```python
#24. Oluşturulan modelin test seti üzerinde doğruluk (accuracy) skorunu hesaplayınız,

print(f'Doğruluk: {randomForest.score(X_test, y_test)}')
```

```python
#25.Değişkenlerin önem derecelerini (Importance) yazdırınız.

print(f'Değişkenlerin İsimleri: {X.columns}\n' + f'Değişkenlerin Önem Dereceleri: {randomForest.feature_importances_}')
```
